## Student data - Mini Project

```python
# count of students
numofstudents = int(input("Total Number Of Students: "))

# data storage
StudentData = []    #it will be stored in the form list & not dictionary

# inserting data
for i in range(numofstudents):
    print(f"Enter the data of Student: {i+1}")   # if use 'i' then it would start from 0
    name = input("Name: ")
    roll_no = int(input("Roll_Number: "))
    marks = int(input("Marks: "))

    if marks > 95:
        grade = "A"
    elif marks >80:
        grade = "B"
    elif marks > 60:
        grade = "C"
    elif marks>= 33:
        grade = "D"
    else:
        grade = "F"

# We will store the data of a student(one student) in a dictionary
    students = {
        "Name": name,
        "Roll_No": roll_no,
        "Marks": marks,
        "Grade": grade
    }

# inserting each students data(student) into one list (Studentdata)
    StudentData.append(students)


# print(StudentData)   will print all the data in a single line, hence cluttered


# making the datas of different students print in different line  [ DATA PRINTING ]
print("All Students Data: \n ")

for s in StudentData:    #will use "loop function" to print all the datas in different line seperately
    print(f"Name: {s['Name']} | Roll Number: {s['Roll_No']} | Marks: {s['Marks']} | Grade: {s['Grade']}")

# PRINTING JUST THE PASSED STUDENTS
print("\n Students who have Passed: ")

for m in StudentData:
    if m['Marks'] >= 33:
        print(f"{m['Name']} - Marks: {m['Marks']} ")

```
